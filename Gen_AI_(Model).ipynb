##**Hugging Face**

###Using API for Hugging face

```python
from google.colab import userdata

api_key = userdata.get("HUGGINGFACEHUB_ACCESS_TOKEN")
```

```python
from langchain_huggingface import ChatHuggingFace,HuggingFaceEndpoint

llm = HuggingFaceEndpoint(
  repo_id="TinyLlama/TinyLlama-1.1B-Chat-v1.0:featherless-ai",
  task="text-generation",
  huggingfacehub_api_token=api_key # Pass the API key here
)

model = ChatHuggingFace(
  llm=llm
)

result = model.invoke("Describe Bihar.")
print(result.content)
```

###Using Locally in our Machine

```python

from langchain_huggingface import ChatHuggingFace,HuggingFacePipeline

llm = HuggingFacePipeline.from_model_id(
  model_id="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
  task="text-generation",
  # The 'featherless-ai' revision was not found, removing the argument.
  # model_kwargs={"revision": "featherless-ai"},
  pipeline_kwargs={
    "max_new_tokens": 100,
    "temperature": 0.5
    }
)

model = ChatHuggingFace(
  llm=llm
)

result = model.invoke("Describe Bihar.")
print(result.content)
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from google.colab import userdata
from sklearn.metrics.pairwise import cosine_similarity

api_key = userdata.get("GEMINI_API_KEY")

embeddings = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-2", # Updated to a currently supported embedding model
    api_key=api_key,
    dimension=32
)
documents = [
    "Virat Kohli is an Indian cricketer known for his aggressive batting and leadership.",
    "MS Dhoni is a former Indian captain famous for his calm demeanor and finishing skills.",
    "Sachin Tendulkar, also known as the 'God of Cricket', holds many batting records.",
    "Rohit Sharma is known for his elegant batting and record-breaking double centuries.",
    "Jasprit Bumrah is an Indian fast bowler known for his unorthodox action and yorkers."
]
query = "Tell me about Modi"
docs_embedings = embeddings.embed_documents(documents)
query_embeding = embeddings.embed_query(query)

scores = cosine_similarity([query_embeding],docs_embedings) #--> it takes 2-D as argument.
#Docs Embedding hmesa 2-D deta hai aur query 1-D isliye query ko 2-D bnane ke liye ek vector aur lgado
idx = scores[0].argmax()
threshold = 0.5
print(scores[0][idx])
if(scores[0][idx]>=threshold):
    print("Score: ",scores[0][idx])
    print("Document: ",documents[idx])
else:
    print("No document found")


```

### **Using Gemini with Temperature**

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")

# Example with a low temperature (more deterministic output)
llm_low_temp = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash", # Updated to a supported model
    api_key=api_key,
    temperature=0.1 # Low temperature
)
result_low_temp = llm_low_temp.invoke("Describe Bihar in one sentence.")
print(f"Response with temperature=0.1: {result_low_temp.text}\n")

# Example with a higher temperature (more creative/diverse output)
llm_high_temp = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash", # Updated to a supported model
    api_key=api_key,
    temperature=0.9 # Higher temperature
)
result_high_temp = llm_high_temp.invoke("Describe Bihar in one sentence.")
print(f"Response with temperature=0.9: {result_high_temp.text}")
```

*  `Same Input` ke liye agr baar baar result chaiye aur `same result` chaiye toh `Temperature` ko `0` rkho
*  `Same Input` ke liye agr baar baar `different result `chaiye aur same result chaiye toh `Temperature` ko `1.5` rkho
